## final_data.csv에서 product_name이 비어 있는 행 확인
백엔드에서 사용하는 `final_data.csv` 기준으로, 이름이 비어 있으면 1순위 등에서 제품명이 안 나올 수 있음.

```python
# product_name이 비어 있거나 NaN인 행
empty_name = df["product_name"].isna() | (df["product_name"].astype(str).str.strip() == "")
n_empty = empty_name.sum()
print(f"product_name 비어 있는 행 수: {n_empty} / {len(df)}")
if n_empty > 0:
    display(df.loc[empty_name, ["product_id", "product_name", "price_krw"]])
else:
    print("비어 있는 행 없음.")
```
